## Phase 2: Target Variable Distribution & Investigation

We analyze the distribution of `buggy` across the training, test, and total datasets to identify class imbalance and dataset construction characteristics.

```python
print('--- Training Set Target Distribution ---')
print(train_df['buggy'].value_counts())
print(train_df['buggy'].value_counts(normalize=True) * 100)

print('\n--- Test Set Target Distribution (Temporal Real-World) ---')
print(test_df['buggy'].value_counts())
print(test_df['buggy'].value_counts(normalize=True) * 100)

print('\n--- Total Dataset Target Distribution ---')
print(total_df['buggy'].value_counts())
print(total_df['buggy'].value_counts(normalize=True) * 100)
```

### Investigation of the `fix` Column
`fix` indicates whether a commit was identified as fixing a bug (via Jira mining).
Notice how many commits that fixed a bug ALSO introduced another bug (regression defects)!

```python
print('Fix column distribution in train_df:')
print(train_df['fix'].value_counts())

print('\nCrosstab: Buggy vs Fix (Train):')
print(pd.crosstab(train_df['buggy'], train_df['fix'], margins=True))

# Why we exclude 'fix' from features: in clean commits, fix was filled with False by definition,
# and in real-time PR review, whether a PR fixes a bug is not consistently known without external linking.
```

## Phase 3: Exploratory Data Analysis (EDA)

We explore numeric distributions, correlations, and how churn relates to defect risk.

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Target distribution visualization
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
train_df['buggy'].value_counts().plot(kind='bar', ax=axes[0], color=['#3b82f6', '#ef4444'], title='Train Set: Artificially Balanced (50/50)')
axes[0].set_xticklabels(['False', 'True'], rotation=0)
axes[0].set_ylabel('Commits')

test_df['buggy'].value_counts().plot(kind='bar', ax=axes[1], color=['#3b82f6', '#ef4444'], title='Test Set: Real-World Distribution (~19% Buggy)')
axes[1].set_xticklabels(['False', 'True'], rotation=0)
axes[1].set_ylabel('Commits')
plt.tight_layout()
plt.show()
```

```python
# Churn vs Buggy status (Log-scale to handle skew)
train_df['total_churn'] = train_df['la'] + train_df['ld']

plt.figure(figsize=(8, 4))
sns.boxplot(x='buggy', y=np.log1p(train_df['total_churn']), data=train_df, palette='Set2')
plt.title('Log(Total Churn) by Buggy Status')
plt.xlabel('Buggy Commit')
plt.ylabel('log1p(la + ld)')
plt.show()
```

### Verify Model Predictions & Explainable Risk Scoring

Load the serialized model artifact and test sample Merge Requests.

```python
import joblib
from src.risk_engine import RiskEngine

artifact = joblib.load('models/mergeguard_model.joblib')
engine = RiskEngine(artifact['pipeline'], artifact['baseline_means'])

# Evaluate an illustrative large PR
mr_sample = {
    'la': 940, 'ld': 230, 'nf': 28, 'nd': 7, 'ns': 3,
    'ent': 3.6, 'ndev': 6, 'age': 120, 'nuc': 30,
    'aexp': 50, 'arexp': 40, 'asexp': 3
}
analysis = engine.analyze(mr_sample, tests_modified=False, critical_modules=True)
print(f'Risk Score: {analysis.risk_score}/100 [{analysis.risk_level}]')
print(f'Defect Probability: {analysis.defect_probability:.2%}')
print('\nTop Risk Factors:')
for f in analysis.top_risk_factors:
    print(f"  - {f['label']}: {f['value']} ({f['description']})")
print('\nRecommendations:')
for r in analysis.recommendations:
    print(f'  - {r}')
```
